```python
from datetime import date, timedelta

from word_api.sources.nyt import get_token, put_puzzle

start = date(2024, 1, 1)
end = date.today()

token = await get_token()

current = start
while current <= end:
    response = await put_puzzle(current, token)
    print(response)
    current += timedelta(days=1)
```
